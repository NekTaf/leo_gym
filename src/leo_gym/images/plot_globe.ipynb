```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from importlib import resources

from leo_gym.utils.matplot_style_cfg import *

with resources.as_file(resources.files("leo_gym") / "images" / "blue_marble.jpg") as p:
    img = imread(p).astype(float) / 255.0

R = 6_371_000.0
lon = np.linspace(0, 2*np.pi, 90) #180
lat = np.linspace(0, np.pi, 45) #90
LON, LAT = np.meshgrid(lon, lat)

x = R*np.cos(LON)*np.sin(LAT)
y = R*np.sin(LON)*np.sin(LAT)
z = R*np.cos(LAT)

H, W = img.shape[:2]
fc = img[(LAT/np.pi*(H-1)).astype(int), (LON/(2*np.pi)*(W-1)).astype(int), :3]

fig = plt.figure(figsize=(3,3))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(x, y, z, facecolors=fc, linewidth=0, antialiased=False, shade=False)

ax.set_box_aspect((1, 1, 1))
plt.tight_layout()
plt.show()
```
